```python
import numpy as np
import pandas as pd
import difflib
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

# Reading the Data

```python
movies = pd.read_csv('data/tmdb_movies_50k.csv')
movies['index'] = movies.index
movies.head()
```

# Filling the Missing Spaces

```python
movies.isnull().sum()
```

```python
selected_features = ['genres', 'keywords', 'overview', 'tagline']
```

```python
for feature in selected_features:
    movies[feature] = movies[feature].fillna('')
```

# Combining the Useful Features

```python
combined_features = movies['genres']+ ' ' +movies['keywords']+ ' ' +movies['overview']+ ' ' +movies['tagline']
combined_features.head()
```

```python
vectorizer = TfidfVectorizer()
feature_vectors = vectorizer.fit_transform(combined_features)
print(feature_vectors)
```

# Cosine Similarity

```python
similarity = cosine_similarity(feature_vectors)
print(similarity)
print(similarity.shape)
```

# Getting the Movie Name from User

```python
movie_name = input('Enter your favourite movie name: ')
```

```python
list_of_all_titles = movies['title'].tolist()
```

```python
find_close_match = difflib.get_close_matches(movie_name, list_of_all_titles)
print(find_close_match)
```

```python
close_match = find_close_match[0]
print(close_match)
```

```python
index_of_the_movie = movies[movies.title == close_match]['index'].values[0]
print(index_of_the_movie)
```

# Finding the Similarity Score

```python
similarity_score = list(enumerate(similarity[index_of_the_movie]))

sorted_similar_movies = sorted(similarity_score, key = lambda x:x[1], reverse=True)
```

# Suggesting the Movies

```python
print("Movies suggested for you : \n")

i = 1

for movie in sorted_similar_movies[1:31]:
    index = movie[0]
    title_of_the_movie = movies[movies.index == index]['title'].values[0]
    if (i<31):
        print(i,'.', title_of_the_movie)
        i += 1
```
